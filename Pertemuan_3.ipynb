```python
# TUGAS PRAKTIKUM 4a
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCVK 2026/peppers.jpg')
original = cv.cvtColor(original, cv.COLOR_BGR2RGB)
image = np.zeros(original.shape, original.dtype)


# Akses per piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        R = int(original[y, x, 0])
        G = int(original[y, x, 1])
        B = int(original[y, x, 2])

        gray = (R + G + B) / 3

        # Nilai grayscale dimasukkan ke semua channel
        image[y, x, 0] = np.clip(gray, 0, 255)
        image[y, x, 1] = np.clip(gray, 0, 255)
        image[y, x, 2] = np.clip(gray, 0, 255)


fig = plt.figure(figsize=(15,15))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(image)
plt.title("Average")

plt.show()
```

```python
# TUGAS PRAKTIKUM 6
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Gamma Correction pada citra')
print('----------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

original = cv.imread('/content/drive/MyDrive/PCVK 2026/houses.JPG')
image = np.zeros(original.shape, original.dtype)

# Akses setiap piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        # OpenCV menggunakan urutan BGR
        B = int(original[y, x, 0])
        G = int(original[y, x, 1])
        R = int(original[y, x, 2])

        # Normalisasi 0-255 menjadi 0-1
        b = B / 255.0
        g = G / 255.0
        r = R / 255.0

        # Gamma Correction
        image[y, x, 0] = np.clip(255 * (b ** gamma), 0, 255)
        image[y, x, 1] = np.clip(255 * (g ** gamma), 0, 255)
        image[y, x, 2] = np.clip(255 * (r ** gamma), 0, 255)

final_frame = cv.hconcat((original, image))
cv2_imshow(final_frame)
```

> Alasan pemilihan: Metode Logarithmic Brightness dipilih karena citra malam memiliki banyak bagian dengan intensitas rendah atau gelap. Oleh karena itu, saya memilih metode perubahan kontras dan brightness untuk memperjelas perbedaan intensitas antara objek dan lingkungan sekitarnya serta mengatur tingkat kecerahan citra. Kedua proses tersebut dapat dilakukan menggunakan satu rumus:

$$ g(x,y)=a \times f(x,y)+b $$

> dengan a sebagai nilai contrast dan b sebagai nilai brightness. Rumus ini juga sesuai dengan transformasi contrast pada modul.

● Menentukan parameter terbaik :
> Parameter terbaik ditentukan dengan melakukan beberapa percobaan terhadap nilai contrast dan brightness. Nilai yang dipilih adalah parameter yang mampu mengurangi kecerahan berlebih serta membuat objek crayfish lebih jelas tanpa menyebabkan bagian citra menjadi terlalu gelap atau terlalu terang.
Parameter terbaik yang saya peroleh adalah contrast = 1.05 dan brightness = -50, karena menghasilkan citra yang lebih jelas dengan tingkat kecerahan yang lebih sesuai dibandingkan citra asli.
