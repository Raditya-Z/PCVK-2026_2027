```python
# TUGAS PRAKTIKUM 2
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print(' Mengubah kontras dan tingkat kecerahan citra ')
print('--------------------------------')
try:
    brightness = int(input('Masukkan tingkat kecerahan: '))
except ValueError:
    print('Error, not a number')

try:
    contrast = float(input('Masukkan kontras: '))
except ValueError:
    print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/sunset.jpg')
image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            image[y,x,c] = np.clip(contrast * original[y,x,c] + brightness, 0, 255)

final_frame = cv.hconcat((original, image))
cv2_imshow(final_frame)
```

```python
# TUGAS PRAKTIKUM 3
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print(' Mengubah tingkat kecerahan citra dengan Tranformasi Log ')
print('--------------------------------')
try:
    lbrightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/houses.JPG')
image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            image[y,x,c] = np.clip(original[y,x,c] * np.log(1 + lbrightness), 0, 255)

final_frame = cv.hconcat((original, image))
cv2_imshow(final_frame)
```

```python
# TUGAS PRAKTIKUM 4a
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCVK 2026/peppers.jpg')
original = cv.cvtColor(original, cv.COLOR_BGR2RGB)
image = np.zeros(original.shape, original.dtype)


# Akses per piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        R = int(original[y, x, 0])
        G = int(original[y, x, 1])
        B = int(original[y, x, 2])

        gray = (R + G + B) / 3

        # Nilai grayscale dimasukkan ke semua channel
        image[y, x, 0] = np.clip(gray, 0, 255)
        image[y, x, 1] = np.clip(gray, 0, 255)
        image[y, x, 2] = np.clip(gray, 0, 255)


fig = plt.figure(figsize=(15,15))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(image)
plt.title("Average")

plt.show()
```

```python
# TUGAS PRAKTIKUM 5
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCVK 2026/peppers.jpg')
original = cv.cvtColor(original, cv.COLOR_BGR2RGB)

# Membuat tempat untuk gambar hasil
image = np.zeros(original.shape, original.dtype)

# Akses per piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        R = int(original[y, x, 0])
        G = int(original[y, x, 1])
        B = int(original[y, x, 2])

        gray = (R + G + B) / 3

        if R > 150 and G < 100 and B < 100:
            image[y, x, 0] = R
            image[y, x, 1] = G
            image[y, x, 2] = B
        else:
            image[y, x, 0] = np.clip(gray, 0, 255)
            image[y, x, 1] = np.clip(gray, 0, 255)
            image[y, x, 2] = np.clip(gray, 0, 255)

# Menampilkan hasil
fig = plt.figure(figsize=(15, 15))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(image)
plt.title("Grayscale dengan Warna Merah")

plt.show()
```

```python
# TUGAS PRAKTIKUM 6
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Gamma Correction pada citra')
print('----------------------------')

try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

original = cv.imread('/content/drive/MyDrive/PCVK 2026/houses.JPG')
image = np.zeros(original.shape, original.dtype)

# Akses setiap piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        # OpenCV menggunakan urutan BGR
        B = int(original[y, x, 0])
        G = int(original[y, x, 1])
        R = int(original[y, x, 2])

        # Normalisasi 0-255 menjadi 0-1
        b = B / 255.0
        g = G / 255.0
        r = R / 255.0

        # Gamma Correction
        image[y, x, 0] = np.clip(255 * (b ** gamma), 0, 255)
        image[y, x, 1] = np.clip(255 * (g ** gamma), 0, 255)
        image[y, x, 2] = np.clip(255 * (r ** gamma), 0, 255)

final_frame = cv.hconcat((original, image))
cv2_imshow(final_frame)
```

```python
# TUGAS PRAKTIKUM 7
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCVK 2026/peppers.jpg', cv.IMREAD_GRAYSCALE)
depth_image = np.zeros(original.shape, original.dtype)

try:
    bit_depth = int(input('Masukkan bit depth tujuan(1-8): '))
except ValueError:
    print('Error, not a number')



level = 255 / (pow(2, bit_depth) - 1)
jumlah_level = pow(2, bit_depth)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        depth_image[y, x] = round(original[y, x]/level) * level

print("Bit depth awal: 8 bit")
print("Bit depth tujuan: " + str(bit_depth) + " bit" )
print("Jumlah level :", jumlah_level)

# Menampilkan hasil
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(original,  cmap= 'gray', vmin=0, vmax=255)
plt.title("Grayscale 8-bit")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(depth_image, cmap='gray', vmin=0, vmax=255)
plt.title("Grayscale " + str(bit_depth) + " bit")
plt.axis("off")

plt.show()
```

```python
# TUGAS PRAKTIKUM 8.1
import cv2 as cv
import numpy as np
import glob
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCVK 2026/galaxy.jpg')

cv_img = []
for img in glob.glob('/content/drive/MyDrive/PCVK 2026/noises/*.jpg'):
    n = cv.imread(img)
    cv_img.append(n)

# Membuat array kosong
average = np.zeros(original.shape, dtype=np.float32)

# Menjumlahkan 10 citra
for i in range(10):
    average = average + cv_img[i]

# Menghitung rata-rata
average = average / 10

# Mengubah menjadi uint8
average = np.clip(average, 0, 255).astype(np.uint8)

# Menghitung PSNR
psnr = cv.PSNR(original, average)

print("Jumlah citra:", 10)
print("PSNR:", psnr, "dB")

# Menampilkan
plt.imshow(cv.cvtColor(average, cv.COLOR_BGR2RGB))
plt.title("Average 10 Citra")
plt.axis("off")
plt.show()
```
