**TUGAS PRAKTIKUM - Analisa**

Pertanyaan : Apakah teks yang ditambahkan pada citra akan ikut terbalik ketika citra di-flip?

Kesimpulan : Berdasarkan percobaan, teks yang ditambahkan sebelum proses flip akan ikut terbalik bersama citra. Hal ini terjadi karena cv.putText() menambahkan teks langsung ke dalam data piksel citra. Ketika cv.flip() dilakukan, seluruh piksel pada citra dibalik, termasuk piksel yang membentuk teks.

```python
img = cv.imread('/content/drive/MyDrive/PCVK 2026/aktivitas.jpg')

cv.putText(img, text='Raditya Zandra Fadhillah', org=(10, 1200), fontFace=cv.FONT_HERSHEY_SIMPLEX, fontScale=1.5, color=(255, 255, 0),thickness=3)

imgFlip = cv.flip(img, 1)

img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
imgFlip = cv.cvtColor(imgFlip, cv.COLOR_BGR2RGB)

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Sebelum Flip")

plt.subplot(1, 2, 2)
plt.imshow(imgFlip)
plt.title("Setelah Flip")

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('/content/drive/MyDrive/PCVK 2026/ktp.jpg')

# Mengambil tinggi dan lebar gambar
tinggi, lebar = img.shape[:2]

print("Ukuran gambar:", lebar, "x", tinggi)

# Menentukan ukuran kanvas persegi
if lebar > tinggi:
    ukuran = lebar
else:
    ukuran = tinggi

# Membuat kanvas berwarna putih
kanvas = np.full((ukuran, ukuran, 3), 255, dtype=np.uint8)

# Menghitung posisi agar gambar berada di tengah
x = int((ukuran - lebar) / 2)
y = int((ukuran - tinggi) / 2) # Perbaikan di sini: tinggi / 2 seharusnya (ukuran - tinggi) / 2

# Memasukkan gambar ke tengah kanvas
kanvas[y:y+tinggi, x:x+lebar] = img

# Menambahkan nama toko
cv.putText(kanvas, text='Toko Raditya', org=(int(ukuran / 20), int(ukuran / 10)), fontFace=cv.FONT_HERSHEY_SIMPLEX, fontScale=ukuran / 1000, color=(0, 0, 0),thickness=int(ukuran / 500))

# BGR ke RGB
kanvas = cv.cvtColor(kanvas, cv.COLOR_BGR2RGB)

# Menampilkan hasil
plt.imshow(kanvas)
plt.title("Thumbnail 1:1")
plt.show()
```
